```python
## clean the data removing special char 
import re
import nltk
from nltk.corpus import stopwords

# Download stopwords once
# nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def clean_text(text):
    # Make sure input is string
    text = str(text)

    # Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Convert to lowercase
    text = text.lower()

    # Remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Remove stopwords
    text = ' '.join(
        word for word in text.split()
        if word not in stop_words
    )

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# Apply cleaning
df['Review'] = df['Review'].apply(clean_text)

# Check result
# print(df[['Review', 'Clean_Review']].head())
```

```python
from gensim.models import Word2Vec

# Tokenize reviews
sentences = [review.split() for review in xtr]

# Train Word2Vec
w2v = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4
)
```

```python
import numpy as np

def document_vector(review, model):
    words = review.split()
    
    vectors = [
        model.wv[word]
        for word in words
        if word in model.wv
    ]
    
    if len(vectors) == 0:
        return np.zeros(model.vector_size)
    
    return np.mean(vectors, axis=0)
```

```python
xtr_w2v = np.array([
    document_vector(review, w2v)
    for review in xtr
])

xte_w2v = np.array([
    document_vector(review, w2v)
    for review in xte
])
```

```python
print(xtr_w2v.shape)
print(xte_w2v.shape)
```

```python
from sklearn.linear_model import LogisticRegression

w2v_model = LogisticRegression(max_iter=1000)

w2v_model.fit(xtr_w2v, ytr)

y_pred = w2v_model.predict(xte_w2v)
```

```python
print(accuracy_score(yte, y_pred))
```

```python
import numpy as np

def predict_review(review, w2v, model):
    # Convert review into Word2Vec document vector
    vector = document_vector(review, w2v)

    # ML model expects 2D input: (1, 100)
    vector = vector.reshape(1, -1)

    # Make prediction
    prediction = model.predict(vector)

    return prediction[0]
```

```python
new_review = "This product is amazing and I really love it"

prediction = predict_review(new_review, w2v, w2v_model)

print(prediction)
```

```python
def predict_review(review, w2v, model):
    vector = document_vector(review, w2v)
    vector = vector.reshape(1, -1)

    prediction = model.predict(vector)[0]
    probability = model.predict_proba(vector)[0]

    return prediction, probability

prediction, probability = predict_review(
    "The product is excellent",
    w2v,
    w2v_model
)

print("Prediction:", prediction)
print("Probability:", probability)
```
